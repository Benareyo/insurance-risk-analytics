### 🔬 Feature Importance and SHAP Explanations in Business Terms

Based on the tree ensembles generated by the Random Forest and XGBoost architectures, we evaluate feature impacts through SHAP value attribution metrics:

1. **`CustomValueEstimate` (Top Influential Vector):** For every additional 10,000 Rand increase in a vehicle's custom asset valuation estimate, the predicted claim severity baseline increases proportionally. This provides hard statistical justification for our value-banded premium tiers.
2. **`VehicleAge`:** Every extra year of vehicle age corresponds to changes in expected repair parts costs. While older vehicles have a lower baseline value, the scarcity of older replacement parts frequently spikes overall claim severity outlays.
3. **`Province_Gauteng`:** Policies linked to the Gauteng metropolitan zone trigger a distinct positive premium adjustment shift, matching the high volume of traffic density accidents recorded during Task 1 EDA analysis.
